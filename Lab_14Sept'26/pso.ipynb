```python

import numpy as np

def rosenbrock(position):
    x, y = position
    return (1 - x)**2 + 100 * (y - x**2)**2

np.random.seed(42)

n_particles = 30
lower_bounds = np.array([-2.0, -1.0])
upper_bounds = np.array([2.0, 3.0])

positions = np.random.uniform(
    lower_bounds, upper_bounds, (n_particles, 2)
)
velocities = np.random.uniform(-0.2, 0.2, (n_particles, 2))

w = 0.7
c1 = 1.5
c2 = 1.5
max_iterations = 100

pbest_positions = positions.copy()
pbest_fitness = np.array([rosenbrock(p) for p in positions])

best_index = np.argmin(pbest_fitness)
gbest_position = pbest_positions[best_index].copy()
gbest_fitness = pbest_fitness[best_index]

for iteration in range(max_iterations):
    r1 = np.random.rand(n_particles, 2)
    r2 = np.random.rand(n_particles, 2)

    velocities = (
        w * velocities
        + c1 * r1 * (pbest_positions - positions)
        + c2 * r2 * (gbest_position - positions)
    )

    positions = positions + velocities
    positions = np.clip(positions, lower_bounds, upper_bounds)

    current_fitness = np.array([rosenbrock(p) for p in positions])

    improved = current_fitness < pbest_fitness
    pbest_positions[improved] = positions[improved]
    pbest_fitness[improved] = current_fitness[improved]

    best_index = np.argmin(pbest_fitness)

    if pbest_fitness[best_index] < gbest_fitness:
        gbest_position = pbest_positions[best_index].copy()
        gbest_fitness = pbest_fitness[best_index]

    if (iteration + 1) % 10 == 0:
        print(
            f"Iteration {iteration + 1}: "
            f"Best Position = {gbest_position}, "
            f"Minimum Value = {gbest_fitness:.6f}"
        )

print("\nFINAL RESULT")
print("Best x, y =", gbest_position)
print("Minimum Function Value =", gbest_fitness)

print("\nTHEORETICAL MINIMUM")
print("Best Position = [1, 1]")
print("Minimum Function Value = 0")
```
