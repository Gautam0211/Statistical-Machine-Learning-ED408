```python

import random

# ==============================
# GENETIC ALGORITHM
# ==============================

# Parameters
POPULATION_SIZE = 8
CHROMOSOME_LENGTH = 6
GENERATIONS = 5
CROSSOVER_POINT = 3
MUTATION_RATE = 0.10
ELITISM = 1

# Objective function
def fitness(n):
    return n**2 + 3*n


# Convert binary chromosome to decimal
def decode(chromosome):
    return int(chromosome, 2)


# Create initial population
def create_population():
    return [
        format(random.randint(0, 63), "06b")
        for _ in range(POPULATION_SIZE)
    ]


# Roulette-wheel selection
def selection(population):
    fitness_values = [
        fitness(decode(chromosome))
        for chromosome in population
    ]

    return random.choices(
        population,
        weights=fitness_values,
        k=1
    )[0]


# Single-point crossover
def crossover(parent1, parent2):
    point = CROSSOVER_POINT

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# Bit-flip mutation
def mutate(chromosome):
    chromosome = list(chromosome)

    for i in range(len(chromosome)):
        if random.random() < MUTATION_RATE:
            chromosome[i] = (
                "1" if chromosome[i] == "0" else "0"
            )

    return "".join(chromosome)


# Genetic Algorithm
def genetic_algorithm():
    population = create_population()

    print("GENETIC ALGORITHM")
    print("Initial Population:", population)

    for generation in range(GENERATIONS):
        population.sort(
            key=lambda c: fitness(decode(c)),
            reverse=True
        )

        best = population[0]

        print(
            f"Generation {generation + 1}: "
            f"Best Chromosome = {best}, "
            f"n = {decode(best)}, "
            f"Fitness = {fitness(decode(best))}"
        )

        # Preserve the best chromosome
        new_population = population[:ELITISM]

        while len(new_population) < POPULATION_SIZE:
            parent1 = selection(population)
            parent2 = selection(population)

            child1, child2 = crossover(parent1, parent2)

            child1 = mutate(child1)
            child2 = mutate(child2)

            new_population.extend([child1, child2])

        population = new_population[:POPULATION_SIZE]

    population.sort(
        key=lambda c: fitness(decode(c)),
        reverse=True
    )

    best = population[0]

    print("\nFinal Best Chromosome:", best)
    print("Best Value of n:", decode(best))
    print("Maximum Fitness:", fitness(decode(best)))


if __name__ == "__main__":
    random.seed(42)
    genetic_algorithm()
```

```python

import random
import math

# ==============================
# SIMULATED ANNEALING
# ==============================

# Parameters
INITIAL_SOLUTION = 2
INITIAL_TEMPERATURE = 20
COOLING_FACTOR = 0.7
MIN_TEMPERATURE = 0.5
MAX_ITERATIONS = 6
STEP_SIZE = 2

# Objective function
def objective(x):
    return x**2 - 8*x + 20


# Simulated Annealing
def simulated_annealing():
    current_solution = INITIAL_SOLUTION
    current_value = objective(current_solution)

    temperature = INITIAL_TEMPERATURE

    print("SIMULATED ANNEALING")

    for iteration in range(MAX_ITERATIONS):

        # Generate a neighboring solution
        candidate = current_solution + STEP_SIZE
        candidate_value = objective(candidate)

        delta = candidate_value - current_value

        # Accept or reject candidate
        if delta < 0:
            accept = True
            probability = 1.0
            random_number = None

        else:
            probability = math.exp(-delta / temperature)
            random_number = random.random()
            accept = random_number < probability

        if accept:
            current_solution = candidate
            current_value = candidate_value

        print(
            f"Iteration {iteration + 1}: "
            f"x = {current_solution}, "
            f"f(x) = {current_value:.2f}, "
            f"T = {temperature:.2f}, "
            f"Accepted = {accept}"
        )

        # Cool the temperature
        temperature *= COOLING_FACTOR

        if temperature < MIN_TEMPERATURE:
            break

    print("\nFinal Solution:", current_solution)
    print("Minimum Function Value:", current_value)


if __name__ == "__main__":
    random.seed(42)
    simulated_annealing()
```
